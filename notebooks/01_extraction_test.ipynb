# Prototype Test 1 — LLM-based Structured Requirement Extraction for Access Control

This notebook contains the first prototype test for extracting structured access control requirements from Dutch natural-language specification text.

The purpose of this prototype is to evaluate whether a large language model can convert textual installation requirements into a structured JSON format that is usable as input for later projection logic within Enginize.

## Scope
- Access control only
- No projection logic
- No time-based logic

## Observations
- Model correctly extracts reader types
- Fails on inheritance in some cases
- Sometimes includes unsupported info in description

```python
requirement_text = """
Alle kantoor deuren moeten voorzien worden van deurdrangers, en in/uit lezers. 
de serverruimte moet voorzien worden worden van een irisscanner
de voordeur moet een pincode krijgen naast de paslezer, inclusief bluetooth.
de parkeergarage krijgt een slagboom met kentekenherkenning en een uitrijlus
de tourniquet bij de receptie moet voorzien worden van een magneetstriplezer
"""
```

```python
import json
from src.parser import extract_requirements_json
from src.schema_access_control import AccessControlSchema

result = json.loads(extract_requirements_json(requirement_text))

try:
    validated = AccessControlSchema.model_validate(result)
    print("Validation: True\n")
    print(json.dumps(validated.model_dump(), indent=2, ensure_ascii=False))
except Exception as e:
    print(f"Validation: False\n{e}")
```
